```python
#12. write a Python program to count total number of notes in given amount.
def count_notes(amount):
    denominations=[2000,500,200,100,50,20,10]
    total_notes=0
    result=[]

    for note in denominations:
        count=amount//note
        if count>0:
            result.append((note,count))
            total_notes+=count
            amount%=note

    return result,total_notes

amount=int(input())
print(amount)
res,total_notes=count_notes(amount)

for note,count in res:
    print(note," ",count)

print("total notes:",total_notes)
```

```python
# 47. Write a python program to check whether a number is Strong number or not.
def strong(n):
    temp = n
    total = 0

    while n > 0:
        dig = n % 10
        fact = 1

        for i in range(1, dig + 1):
            fact = fact * i

        total = total + fact
        n = n // 10

    if total == temp:
        return True
    else:
        return False


n = int(input())
print(n)

res = strong(n)

if res:
    print("Strong number")
else:
    print("Not Strong number")
        
```

```python
# 79. WAP to read numbers from a file and write even, odd and prime numbers to separate file.
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def separate_numbers(filename):
    f = open(filename, "r")
    numbers = f.read().split()
    f.close()

    even_file = open("even.txt", "w")
    odd_file = open("odd.txt", "w")
    prime_file = open("prime.txt", "w")

    for i in numbers:
        n = int(i)

        if n % 2 == 0:
            even_file.write(str(n) + " ")

        else:
            odd_file.write(str(n) + " ")

        if is_prime(n):
            prime_file.write(str(n) + " ")

    even_file.close()
    odd_file.close()
    prime_file.close()

    return True


filename = input()

res = separate_numbers(filename)

if res:
    print("Numbers separated successfully")
```

```python
# 83. WAP to check whether a number is prime, Armstrong or perfect number using functions.
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True


def is_armstrong(n):
    temp = n
    total = 0
    count = len(str(n))

    while n > 0:
        digit = n % 10
        total = total + digit ** count
        n = n // 10

    if total == temp:
        return True
    else:
        return False


def is_perfect(n):
    total = 0

    for i in range(1, n):
        if n % i == 0:
            total = total + i

    if total == n:
        return True
    else:
        return False


n = int(input())

if is_prime(n):
    print("Prime number")
else:
    print("Not a Prime number")

if is_armstrong(n):
    print("Armstrong number")
else:
    print("Not an Armstrong number")

if is_perfect(n):
    print("Perfect number")
else:
    print("Not a Perfect number")
```

```python
# 85. WAP to print all strong numbers between given interval using functions.
def is_strong(n):
    temp = n
    total = 0

    while n > 0:
        digit = n % 10
        fact = 1

        for i in range(1, digit + 1):
            fact = fact * i

        total = total + fact
        n = n // 10

    if total == temp:
        return True
    else:
        return False


def strong_numbers(start, end):
    result = []

    for i in range(start, end + 1):
        if is_strong(i):
            result.append(i)

    return result


start = int(input())
end = int(input())

res = strong_numbers(start, end)

for i in res:
    print(i, end=" ")
```

```python
# 89. WAP to Display the ATM Transaction
def atm_transaction(balance, amount):
    if amount <= 0:
        return balance, "Invalid amount"

    if amount > balance:
        return balance, "Insufficient balance"

    balance = balance - amount
    return balance, "Transaction successful"


balance = float(input())
amount = float(input())

balance, message = atm_transaction(balance, amount)

print(message)
print("Remaining balance:", balance)
```

```python
# 90. WAP to print reverse case of a given string
def reverse_case(s):
    result = ""

    for i in s:
        if i.isupper():
            result = result + i.lower()
        elif i.islower():
            result = result + i.upper()
        else:
            result = result + i

    return result


s = input()

res = reverse_case(s)

print("Reverse case:", res)
```
